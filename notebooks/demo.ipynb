## 分类方法

### 读取数据集

```python
##读取数据集
dataTrain = pd.read_csv(r'Train_samples.csv')
print('数据集结构：', dataTrain.shape)
print('变量维度：', len(dataTrain.columns))
dataTrain = dataTrain.rename(columns={'fpd4':'label'})
#选择可用的变量空间
feaSpace = [i for i in dataTrain.columns if i not in ['create_month', 'label', 'cust_no']]
```

### 工具演示

#### 决策树方法

```python
result=atsr_rule_finder_clf(data=dataTrain, n_rounds=20, num_feature_min=2, num_feature_max=5, params=feaSpace[:20], label='label', depth=7, min_samples_leaf=100, random_state=None)
```

```python
df_reports = atsr_report_clf(dataTrain, result, rule_col='rule_str', label='label', groups='create_month', id_col='cust_no')
df_reports.head()
```

```python
print(atsr_rule_finder_clf.__doc__)
```

```python
print(atsr_report_clf.__doc__)
```

#### xgboost方法

```python
result = atsr_rule_finder_clf_xgb(dataTrain, n_rounds=100, params=feaSpace, label='label', depth=7, min_samples_leaf=50, random_state=None)
```

```python
df_reports = atsr_report_clf(dataTrain, result, rule_col='rule_str', label='label', groups='create_month', id_col='cust_no')
df_reports.head()
```

## 回归方法

### 读取数据集

```python
##读取数据集
dataTrain  = pd.read_csv('OOT_sample_reg.csv')
print('数据集结构：', dataTrain.shape)
print('变量维度：', len(dataTrain.columns))
dataTrain = dataTrain.rename(columns={'profit' : 'cost'})
feaSpace = [i for i in dataTrain.columns if i not in ['create_month', 'profit', 'cust_no']]
```

### 工具演示

#### 决策树方法

```python
result = atsr_rule_finder_reg(dataTrain, n_rounds=10, num_feature_min=2, num_feature_max=5, params=feaSpace, 
                              label='cost', depth=7, min_samples_leaf=1000)
```

```python
df_reports = atsr_report_reg(dataTrain, result, rule_col='rule_str', label='cost', groups='create_month', id_col='cust_no')
df_reports.head()
```

```python
df_reports = atsr_report_reg(dataTrain, result, rule_col='rule_str', label='cost', groups='create_month', id_col='cust_no')
df_reports.head()
```
